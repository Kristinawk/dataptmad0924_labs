#### Display the trending Python repositories in GitHub.

The steps to solve this problem is similar to the previous one except that you need to find out the repository names instead of developer names.

````python
# This is the url you will scrape in this exercise
url = 'https://github.com/trending/python?since=daily'
````

````python
# 1. Hacemos el request y lo instanciamos
response = requests.get(url)
response
````

````python
# 2. Instanciamos el contenido del request
html=response.content
````

````python
# 3. Parseamos el contenido del request
parsed_html=bs4.BeautifulSoup(html, "html.parser")
parsed_html
````

````python
# 4. Nos fijamos qué características tiene el tag que contiene el nombre del repositorio y lo instanciamos
article = parsed_html.find_all("a", {"class": "Link", "href": True, "data-hydro-click": True})
````

````python
article
````

````python
# 5. Exploramos los datos para ver dónde está excatamente la información que necesitamos
print(article[0])
print(type(article[0]))
````

````python
# 6. Accedemos al contenido que nos interesa
article[0]["href"]
````

````python
#7. Hacemos una list comprehension para acceder a todos los repositorios
````

````python
trending_repositories=[article[i]["href"] for i in range(len(article))]
trending_repositories
````
