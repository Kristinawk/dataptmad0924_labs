#### From the Fitbit data, we want to visually understand:

How the average number of steps change by month. Use the appropriate visualization to show the median steps by month. Is Fitbitter more active on weekend or workdays? All plots must be in the same jupyter notebook cell.

**Hints**:

* Use Months_encoded and Week_or Weekend columns.
* Use matplolib.pyplot object oriented API.
* Set your size figure to 12,4
* Explore plt.sca
* Explore plt.xticks
* Save your figures in a folder called `figures` in your repo. 

```python
# MEDIAN STEPS BY MONTH_ENCODED
# your code here

# Group dataset by month:
df_steps_month = fitbit.groupby(['Months_encoded'])[['Steps']].mean().reset_index()

# Create a list of month names
month_names = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 
               'September', 'October', 'November', 'December']

# Map the 'Months_encoded' to month names
df_steps_month['Month_Name'] = df_steps_month['Months_encoded'].apply(lambda x: month_names[x-1])

# Create the plot
ax1 = df_steps_month.plot(x='Month_Name', y='Steps', kind='bar', figsize=(12,4))

# Set x-axis label and title
ax1.set_xlabel('Month')
ax1.set_ylabel('Average Steps')
ax1.set_title('Average Steps per Month')

# Set minimum value for y-axis
plt.ylim(bottom=5000)

# Save plot
plt.savefig('../Data/plot1.png', transparent=False, dpi=80, bbox_inches="tight", format='png')


# MEDIAN STEPS BY WORK_OR_WEEKEND

# Group dataset by week:
df_steps_weekday = fitbit.groupby(['Work_or_Weekend'])[['Steps']].mean().reset_index()

# Create the plot
ax2 = df_steps_weekday.plot(x='Work_or_Weekend', y='Steps', kind='bar', figsize=(12,4))

# Change the name of x-axis labels
plt.xticks([0, 1], ["Weekend" , "Work"])

# Set x-axis label and title
ax2.set_xlabel('Wekday_type')
ax2.set_ylabel('Average Steps')
ax2.set_title('Average Steps per Weekday_type')

# Set minimum value for y-axis
plt.ylim(bottom=10000)

# Save plot
plt.savefig('../Data/plot2.png', transparent=False, dpi=80, bbox_inches="tight", format='png')
```

#### Write a loop to plot 3 scatter plots of the following features:

* Minutes Lightly Active vs Steps    
* Minutes Very Active vs Steps    
* Minutes Sedentary vs Steps  

```python
# your code here
Minuts_Active_list = ["Minutes Lightly Active", "Minutes Fairly Active", "Minutes Very Active"]
for column in Minuts_Active_list:
    fitbit.plot(x=column, y='Steps', kind='scatter', figsize=(12,4))
    plt.xlabel(column)
    plt.ylabel('Steps')
    plt.title(f'{column} vs Steps')
```
