#### Use your previous code but now, resize your plot.
**Hint**: Add the figsize argument in plt.subplots().

If you want, try to add a title to the plot or even axes labels. You can also play with the fontweight and fontsize of the titles and labels. 

```python
# your code here
fig, axs = plt.subplots(ncols = 2,
                       nrows = 1,
                       figsize = (10, 5))
axs[0].plot(x, y)                    
axs[1].plot(x, z)

axs[0].set(xlabel='x',
           ylabel='y',
           title='Linear growth')
axs[1].set(xlabel='x',
           ylabel='z',
           title='Exponential growth')

# Alternative to set labels and title, with formatting

axs[0].set_title('Linear growth', fontsize=10, fontweight='bold')
axs[0].set_xlabel('X', fontsize=10, fontweight='light')
axs[0].set_ylabel('Y', fontsize=10, fontweight='light')

axs[1].set_title('Exponential growth', fontsize=10, fontweight='bold')
axs[1].set_xlabel('X', fontsize=10, fontweight='light')
axs[1].set_ylabel('Z', fontsize=10, fontweight='light')

# Adjust the space between subplots
plt.subplots_adjust(wspace=0.4)
```
